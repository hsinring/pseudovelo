# PseudoVelo Case 4 - Comprehensive Evaluation and Robustness Analysis

```python
import warnings, sys, os
import logging

def customwarn(message, category, filename, lineno, file=None, line=None):
    pass
warnings.showwarning = customwarn
warnings.filterwarnings("ignore")
warnings.filterwarnings("ignore", message="No data for colormapping provided via 'c'.*")
warnings.filterwarnings("ignore", category=DeprecationWarning)
logging.getLogger().setLevel(logging.ERROR) 
os.environ['PYTHONWARNINGS'] = 'ignore'

import scanpy as sc
sc.settings.verbosity = 0 
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import pseudovelo as pv
import cellrank as cr
from cellmapper import CellMapper
from cellrank.kernels import CytoTRACEKernel
from transition import evaluate_transition_accuracy, projection_consistency

np.random.seed(0)
```

```python
def plot_projection_consistency_boxplot(data, labels=None):
    plt.figure(figsize=(2.5, 2.5))
    if isinstance(data, pd.Series) or (isinstance(data, np.ndarray) and data.ndim == 1):
        data = [data]
    
    sns.boxplot(data=data, width=0.2, palette="Set2", fliersize=2)
    for i, d in enumerate(data):
        plt.text(i + 0.15, np.nanmedian(d), f'{np.nanmedian(d):.2f}', ha='left', va='center', fontsize=8)
    
    plt.ylabel('Projection Consistency', fontsize=10)
    if labels is not None:
        plt.xticks(range(len(labels)), labels, fontsize=10)
    else:
        plt.xticks(fontsize=10, rotation=15)
    sns.despine()
    plt.tight_layout()
    plt.show()

def load_and_preprocess_data(subsample_cells=False, subsample_genes=False):
    adata = sc.read_h5ad("./data/bone/human_cd34_bone_marrow.h5ad")
    
    if subsample_cells:
        sc.pp.subsample(adata, fraction=0.5)
    if subsample_genes:
        adata = adata[:, np.random.choice(adata.n_vars, size=adata.n_vars // 2, replace=False)].copy()
        
    sc.pp.filter_genes(adata, min_cells=10)
    sc.pp.normalize_total(adata)
    sc.pp.log1p(adata)
    sc.pp.highly_variable_genes(adata, n_top_genes=2000)
    adata = adata[:, adata.var.highly_variable].copy()
    sc.tl.pca(adata)
    sc.pp.neighbors(adata, n_neighbors=30, n_pcs=30)
    return adata

def get_safe_root(adata, target_idx, fallback_cluster="HSC_1", cluster_key="clusters"):
    if target_idx < adata.n_obs:
        return target_idx
    idx = np.where(adata.obs[cluster_key] == fallback_cluster)[0]
    return np.random.choice(idx)

def run_eval_pipeline(adata, time_key="dpt_pseudotime", cluster_key="clusters", run_dpt=True, root_idx=None, ctk=None, n_states=6):
    
    # 1. DPT Root
    if run_dpt and root_idx is not None:
        adata.uns["iroot"] = root_idx
        sc.tl.diffmap(adata)
        sc.tl.dpt(adata)
        fig, ax = plt.subplots(figsize=(3.7,3))
        sc.pl.embedding(adata, color="dpt_pseudotime", show=False, basis="tsne", ax=ax, frameon=False, size=15)
        start_x, start_y = adata.obsm["X_tsne"][root_idx, 0], adata.obsm["X_tsne"][root_idx, 1]
        plt.scatter(start_x, start_y, color='red', marker='*', s=200, edgecolors='black', zorder=5)
        plt.show()

    # 2. PseudoVelo
    adata.layers["X"] = adata.X.toarray()
    pv.fit_nb_gam_with_center_diff(adata, pseudotime_key=time_key, n_jobs=16, cluster_key=cluster_key, input_layer="X", reverse=False)
    pv.plot_velocity_projection(adata, xkey="X", vkey="pseudo_velocity", basis="tsne", color="clusters", 
                                figsize=(3.9,3.1), title="PseudoVelo", size=120, legend_loc="none")

    # 3. CellRank2
    if time_key == "dpt_pseudotime":
        kernel = cr.kernels.PseudotimeKernel(adata, time_key=time_key)
        kernel.compute_transition_matrix()
        kernel.write_to_adata()
        
        kernel.plot_projection(basis="tsne", color="clusters", show=False, figsize=(3.9,3.1), 
                            title="CellRank2", s=120, legend_fontsize=8, legend_loc="none")
        plt.gca().grid(False)
        plt.show()
    else:
        kernel = ctk

    # 4. Quantitative Analysis 1: Consistency
    c_pv = projection_consistency(adata, embedding_key='X_tsne', vkey='pseudo_velocity', l2_norm=False)
    c_cr2 = projection_consistency(adata, embedding_key='X_tsne', l2_norm=False, T_velo=kernel.transition_matrix)
    adata.obs[f'consistency_{time_key}_pv'] = c_pv
    adata.obs[f'consistency_{time_key}_cr2'] = c_cr2
    plot_projection_consistency_boxplot([c_pv, c_cr2], labels=["PseudoVelo", "CellRank2"])

    # 5. Quantitative Analysis 2: Transition Accuracy
    order = [['HSC_1', 'HSC_2', 'Precursors', 'Mono_2', 'Mono_1'], ['HSC_1', 'Ery_1', 'Ery_2']]
    acc_pv = evaluate_transition_accuracy(adata, cluster_key="clusters", lineage_order=order, vkey='pseudo_velocity')
    acc_cr2 = evaluate_transition_accuracy(adata, cluster_key="clusters", lineage_order=order, T=kernel.transition_matrix)
    print(f"{time_key} - pv accuracy:", acc_pv)
    print(f"{time_key} - cr2 accuracy:", acc_cr2)

    # 6. GPCCA Analysis
    g_cr2 = cr.estimators.GPCCA(kernel)
    g_cr2.compute_macrostates(cluster_key="clusters", n_states=n_states)
    g_cr2.predict_initial_states()
    g_cr2.predict_terminal_states(allow_overlap=True)
    g_cr2.plot_macrostates(which="initial", legend_loc="on data", s=100, basis="tsne", figsize=(4,4))
    g_cr2.plot_macrostates(which="terminal", legend_loc="on data", s=100, basis="tsne", figsize=(4,4))

    vk = cr.kernels.VelocityKernel(adata, vkey="pseudo_velocity", xkey="X")
    vk.compute_transition_matrix()
    g_pv = cr.estimators.GPCCA(vk)
    g_pv.compute_macrostates(n_states=n_states, cluster_key="clusters")
    g_pv.predict_initial_states()
    g_pv.predict_terminal_states(allow_overlap=True)
    g_pv.plot_macrostates(which="initial", legend_loc="on data", s=100, basis="tsne", figsize=(4,4))
    g_pv.plot_macrostates(which="terminal", legend_loc="on data", s=100, basis="tsne", figsize=(4,4))
    
    return adata, kernel
```

### Part 1: Standard Evaluation and CytoTRACE Comparison

```python
# 1.1 DPT Standard
adata_dpt = load_and_preprocess_data()
coords = adata_dpt.obsm["X_tsne"][:, 0:2]
hsc1_idx = np.where(adata_dpt.obs["clusters"] == "HSC_1")[0]
root_idx = hsc1_idx[np.argmax(coords[hsc1_idx, 1])]
adata_dpt, pk = run_eval_pipeline(adata_dpt, time_key="dpt_pseudotime", cluster_key="clusters", root_idx=root_idx, n_states=6)
```

```python
# 1.2 CytoTRACE Analysis
adata_raw = sc.read_h5ad("./data/bone/human_cd34_bone_marrow.h5ad")
adata_ct = adata_raw.copy()

sc.pp.filter_genes(adata_ct, min_cells=10)
sc.pp.normalize_total(adata_ct)
sc.pp.log1p(adata_ct)
sc.pp.highly_variable_genes(adata_ct, n_top_genes=2000)
sc.tl.pca(adata_ct)
sc.pp.neighbors(adata_ct, n_neighbors=30, n_pcs=30)

adata_ct.layers["imputed"] = CellMapper(adata_ct).map(layer_key="X", use_rep="X_pca").query_imputed.X.copy()
ctk = CytoTRACEKernel(adata_ct).compute_cytotrace(layer="imputed")
adata_ct.obs["1_minus_pseudotime"] = 1 - adata_ct.obs["ct_pseudotime"]

fig, ax = plt.subplots(figsize=(3.7,3))
sc.pl.embedding(adata_ct, basis="tsne", color="1_minus_pseudotime", frameon=False, ax=ax, size=15, title="CytoTRACE score", legend_fontsize=8)
ctk.compute_transition_matrix()
ctk.plot_projection(basis="tsne", color="clusters", figsize=(3.9,3.1), title="CellRank2", size=120, legend_loc="none", show=False)
ax = plt.gca()
ax.grid(False)
plt.show()

# Build adata for PV (align genes and parameters)
adata_pv_ct = load_and_preprocess_data()
adata_pv_ct.obs["ct_pseudotime"] = adata_ct.obs["ct_pseudotime"]
adata_pv_ct, ctk = run_eval_pipeline(adata_pv_ct, time_key="ct_pseudotime", cluster_key="clusters", run_dpt=False, ctk=ctk, n_states=6)
```

### Part 2: Robustness Analysis
Test the impact of re-clustering, cell subsampling, and gene subsampling on the results.

```python
# 2.1 Re-clustering Test (Leiden resolution 1.0)
print("\n--- Running Robustness: Re-clustering (Leiden) ---")
adata_rob_cluster = load_and_preprocess_data()
sc.tl.leiden(adata_rob_cluster, resolution=1.0)
fig, ax = plt.subplots(figsize=(3,3))
sc.pl.embedding(adata_rob_cluster, color="leiden", basis="tsne", ax=ax, frameon=False)
coords = adata_rob_cluster.obsm["X_tsne"][:, 0:2]
hsc1_idx = np.where(adata_rob_cluster.obs["clusters"] == "HSC_1")[0]
root_idx = hsc1_idx[np.argmax(coords[hsc1_idx, 1])]
_ = run_eval_pipeline(adata_rob_cluster, cluster_key="leiden", root_idx=root_idx, n_states=6)
```

```python
# 2.2 Cell Subsampling Test
print("\n--- Running Robustness: Cell Subsampling ---")
adata_rob_cell = load_and_preprocess_data(subsample_cells=True)
sc.tl.diffmap(adata_rob_cell)
coords = adata_rob_cell.obsm["X_tsne"][:, 0:2]
hsc1_idx = np.where(adata_rob_cell.obs["clusters"] == "HSC_1")[0]
target_idx = hsc1_idx[np.argmax(coords[hsc1_idx, 1])]
root_idx=target_idx
print(root_idx)
_ = run_eval_pipeline(adata_rob_cell, root_idx=root_idx, cluster_key="clusters", n_states=6)
```
